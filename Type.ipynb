```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11)
x
y=x**2
y
```

```python
fig=plt.figure(figsize=(20,8),dpi=100)
axis1=fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.4,0.2,0.2])
axis2.plot(x,y)
```

Type of Plot

```python
#scatter plot
plt.scatter(x,y)
```

```python
#histogram
from random import sample
data=sample(range(1,1000),100)
plt.hist(data)
```

```python
#boxplot
# data1=sample(range(1,1000),100)
plt.boxplot(x)
```

Saving Plot

```python
fig=plt.figure()
ax=fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig('mygraph.png',dpi=200)
```
